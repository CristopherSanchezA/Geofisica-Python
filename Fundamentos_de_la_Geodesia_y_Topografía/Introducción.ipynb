**Funciones: def**

```python
# Definir la función
def multiplicar(a, b=2):
    """
    Esta función multiplica dos números.
    Si no se da el segundo número, por defecto se multiplica por 2.
    """
    return a * b
```

```python
# Ejemplos de uso
resultado1 = multiplicar(3, 4)   # Multiplica 3 * 4
resultado2 = multiplicar(5)      # Multiplica 5 * 2 (usa el valor por defecto)
resultado3 = multiplicar(a=7, b=6)  # Uso con argumentos nombrados

# Mostrar resultados
print("3 * 4 =", resultado1)
print("5 * 2 =", resultado2)
print("7 * 6 =", resultado3)
```

Probemos utilizar un booleano con la función multiplicar

- Numpy y matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np

# Definimos tiempo 
t = np.arange(0, 60, 0.05)  # 60 segundos con paso de 0.05 s

# Señal simulada: combinación de ondas + decaimiento exponencial
onda1 = np.sin(2 * np.pi * 0.5 * t) * np.exp(-0.02 * t)   # onda de baja frecuencia
onda2 = 0.5 * np.sin(2 * np.pi * 2.0 * t) * np.exp(-0.05 * t)  # onda más rápida
ruido = 0.1 * np.random.randn(len(t))  # ruido aleatorio

# Sismograma sintético (onda1 + onda2 + ruido)
sismograma = onda1 + onda2 + ruido

# Graficar
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(t, sismograma, color="black", linewidth=1)

# Personalizar
ax.set(
    xlabel="Tiempo (s)",
    ylabel="Amplitud (m/s)",
    title="Ejemplo de sismograma sintético"
)
ax.grid(True)

plt.show()
```
