## ¿Por qué estudiar los sistemas de referencia?
En geodesia, uno de los principales objetivos es la determinación de la posición de un punto sobre la superficie terrestre. Para eso, necesitamos un sistema de referencia que nos permita ubicar cualquier lugar en la Tierra de manera consistente.

## ¿Esfera, Elipsoide o Geoide?
- La esfera es el modelo más simple
- El elipsoide es como una esfera ‘achatada’, que se aproxima mejor a la forma real de la Tierra.
- El geoide es el modelo más detallado (no real) porque incorpora las irregularidades del campo gravitatorio y coincide con el nivel medio del mar extendido bajo los continentes. ¿Qué significa que sea equipotencial?

### Coordenadas geográficas
- Son las más usadas a nivel global y se basan en un modelo esférico de la Tierra.
- Se basan en latitud y longitud:
  - Latitud (ϕ): Mide la distancia hacia el norte o sur del ecuador, en grados. Varía de -90° (Polo Sur) a +90° (Polo Norte).
  - Longitud (λ): Mide la distancia hacia el este o oeste del meridiano de referencia (el meridiano de Greenwich), también en grados. Varía de -180° a +180°.
- **Ejemplo:** 36.827° S, 73.0498° O corresponde a Concepción.
- También se pueden expresar en grados decimales.

### Coordenadas cartesianas
- Otro sistema es el cartesiano, que usa tres ejes: X, Y y Z, con origen en el centro de la Tierra.
- El eje Z apunta a los polos, el X al meridiano de Greenwich, y el Y hacia el este en el ecuador.
- Podemos transformar coordenadas cartesianas en esféricas y viceversa usando fórmulas o código en Python.

Las coordenadas cartesianas (x,y,z) y las coordenadas esféricas (ρ,θ,φ) de un punto se relacionan de la siguiente manera:

### Coordenadas esféricas a cartesianas
x = ρsinφcosθ

y = ρsin⁡φsin⁡θ

z = ρcosφ
- ρ: es la distancia radial desde el origen al punto (como el “radio”).
- θ: es el ángulo azimutal, se mide en el plano XY desde el eje X positivo (similar a la longitud).
- φ: es el ángulo polar, se mide desde el eje Z positivo hacia abajo (similar a la colatitud).

### Coordenadas cartesianas a esféricas
ρ² = x²+y²+z²

tanθ = x/y​

φ = arccos(x/sqrt(x²+y²+z²))
- ρ: se obtiene como la raíz de la suma de cuadrados de x, y y z (distancia al origen).
- θ: se obtiene con el cociente y/x, que indica el ángulo en el plano horizontal.
- φ: depende de la altura z en relación al radio total.
