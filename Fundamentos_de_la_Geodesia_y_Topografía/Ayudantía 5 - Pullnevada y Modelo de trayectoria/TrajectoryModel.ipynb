Modelo de Bevis and Brown 2014

1. Componente lineal: modelo de velocidad cte, podemos describir el movimiento de la tierra como una recta.

2. Heaviside (Saltos): Modela cambios instantáneos (terremoto, cambio de antena).

3. Componente estacional (Anual y semianual): captura variaciones periódicas.

4. Aceleraciones: existen aceleraciones, consideramos un término polinomlial para representarla.

5. Deformación postsísmica: SLTM no funciona para modelar deformación postsísmica, asumimos una formula transitoria logarítmica.
