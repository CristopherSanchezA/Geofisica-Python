¿Qué es interpolacipon y regresión?
Se denomina interpolación a la obtención de nuevos puntos partiendo del conocimiento de un conjunto discreto de puntos. La regresión es el ajuste de los datos experimentales a una función. Por ejemplo, los datos experimentales (timepo, precipitación) se podrían ajustar a una línea recta, que no tiene por qué pasar poo todos los puntos. Pero en la interpolación, teóricamente la función debe pasar por todos los puntos.

Sobreajuste: Es ocupar por ejemplo interpolación, forzando a pasar por todos los puntos pero el resultado es "falso" o no representaría la verdad de los datos.

Robustez y Resistencia en estadística: Robusto es cuando algo se logra aproximar a una cosa, una serie está bien aproximada con esta función, eso quiere decir que serie es robusta. Resistencia es que los datos extremos o "outliers" no logran que tu función aproximada se vea alterada por estos datos inesperados.

La mediana es resistente, ya que no se ve afectada por datos en los bordes, depende sólo del número de datos, no del valor de estos.

Regresión Lineal: Método que madela la relación entre dos variables, la independiente (x) y la dependiente (y) a través de una ecuación lineal. Es una recta que NO pasa por todos los datos, sólo nos da una representación sencilla. Existen dos métodos o formas de definir una regresión lineal, la simple que está dada por: y = \beta_1 x + \beta_0 + \epsilon(ecuación de la recta), donde \beta_1: pendiente, \beta_0: condición inicial (y(0)), \epsilon: error.

El método que vamos a usar para definir la regresión lineal y construir la ecuación de la recta es la de mínimos cuadrados lineales, esta técnica o criterio se utiliza para encontrar la mejor curva (ecuación) para los datos, minimizando el error asociado.
